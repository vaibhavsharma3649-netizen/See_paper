```python
df.drop(columns=['summary_word_count','published_date','updated_date','first_author'],inplace=True)
```

```python

df["tags"] = (
    df["title"] + " " +
    df["summary"] + " " +
    df["authors"]
)
```

```python
df["tags"] = df["tags"].str.lower()
```

```python
df_small = df.sample(
    n=100000,
    random_state=42
).reset_index(drop=True)
```

```python
df_small.head()
```

# Text Vectorization

```python
#using text vectorisation
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2",device='cuda')
```

```python
df_small["tags"] = df_small["tags"].str.lower()
```

```python
vectors=model.encode(df_small['tags'].values.tolist(),show_progress_bar=True)
```

```python
print(vectors)
```

```python
import chromadb
```

```python
client = chromadb.PersistentClient(
    path="/kaggle/working/chroma_db_v3"
)
```

```python
collection = client.get_or_create_collection(
    name="research_papers_v3"
)
```

```python
df_small["arxiv_id"] = (
    df_small["id"]
    .astype(str)
    .str.replace(r"^abs-", "", regex=True)
)
```

```python
batch_size = 5000


collection.delete(ids=[str(i) for i in range(len(df_small))])

for start in range(0, len(df_small), batch_size):

    end = min(start + batch_size, len(df_small)) # Ensure we use len(df_small) here

    collection.add(
        ids=[str(i) for i in range(start, end)],
        embeddings=vectors[start:end].tolist(),
        documents=df_small["tags"].iloc[start:end].tolist(),
        metadatas=[
            {
                "title": row["title"],
                "category_code": row["category_code"],
                "arvix_id":row['arxiv_id']
            }
            for _, row in df_small.iloc[start:end].iterrows() # Ensure we use df_small.iloc here
        ]
    )

    print(f"Added {start} to {end}")
```

```python
# 1. Write your search query
query = "deep learning "

# 2. Convert query into an embedding
query_vector = model.encode([query])

# 3. Search ChromaDB
results = collection.query(
    query_embeddings=query_vector.tolist(),
    n_results=5
)

# 4. Display results
# Add a check to ensure results and metadatas are not None before iterating
if results and results.get("metadatas") and len(results["metadatas"]) > 0 and results["metadatas"][0]:
    print("Found relevant papers:")
    for i, metadata in enumerate(results["metadatas"][0]):
        if metadata: # Ensure metadata item is not None
            print(f"{i+1}. {metadata['title']}")
        else:
            print(f"{i+1}. [Metadata missing for this result]") # Handle cases where individual metadata items might be None
else:
    print("No relevant papers found for the query or an issue occurred during retrieval.")
```

```python
df_small["arxiv_id"] = df_small["id"].astype(str)
df_small.drop(columns=["id"], inplace=True)

df_small.to_csv(
    "/kaggle/working/papers.csv",
    index=False
)
```
